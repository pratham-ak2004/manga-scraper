```python
import re

def extract_chapter_number(name):
    # Match a number (possibly decimal) at the end of the string
    match = re.search(r'(\d+(?:\.\d+)?)\s*$', name)
    if match:
        return match.group(1)
    return None

# Examples:
print(extract_chapter_number("chapter 1"))        # 1
print(extract_chapter_number("chapter 2.1"))      # 2.1
print(extract_chapter_number(".... 5"))           # 5
print(extract_chapter_number("random text 10.5")) # 10.5
print(extract_chapter_number("vol 3 extra"))      # None
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(html_content, 'html.parser')
```

```python
details = soup.find('main').find_all('section')[1]
```

```python
name = details.find_next('h1').text
picture = details.find('picture').find_next('img')['src']
```

```python
list_items = details.find("ul").find_all("li")
```

```python
details_map = {}

for li in list_items:
    key = li.find_next('strong').text.strip()
    values = li.find_all('a')
    value_list = list()
    for val in values:
        value_list.append({
            "link": val["href"],
            "text": val.text or ""
        })
    
    details_map[key[:-1]] = value_list
    
```

```python
import os

def has_png_files(root_folder):
    for root, dirs, files in os.walk(root_folder):
        for file in files:
            if file.lower().endswith(".png"):
                return True
    return False

if __name__ == "__main__":
    folder_path = "./data"

    if has_png_files(folder_path):
        print("PNG files found.")
    else:
        print("No PNG files found.")
```

```python
import requests
from PIL import Image
from io import BytesIO
```

```python
img_url = "https://hot.planeptune.us/manga/Vinland-Saga/0020-018.png"
file_path = "test.png"
```

```python
response = requests.get(img_url, timeout=60)
response.raise_for_status()
```

```python
with open(file_path, 'wb') as file:
    img = Image.open(BytesIO(response.content))
    if img.mode not in ('RGB', 'RGBA'):
        img = img.convert('RGB')
    img.save(file, format='PNG')
```
